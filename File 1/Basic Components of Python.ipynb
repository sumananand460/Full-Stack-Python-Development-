## Inheritance:

### Inheritance means passing on of existing features to new class
<pre>
    <b>
        1. In inheritance the features are always passed from the top class to the bottom class.
        2. In order to implement the concept we required minimum two classes.
        3. Inheritance follows IS-A relationship.
        4. The class who gives it's properties to another class is known as parent class/superclass/base class.
        5. The class who receives the properties from the parent class is known child class/sub-class/drived class.
        6. In inheritance concept in order to access the classes members present in the parent class and child class we must create object for the child class.
</pre>
<pre>
    <b>
        Types of Inheritance:

        In python we have 3 Inheritance:

        1. Single Inheritance
        2. Multi-Level Inheritance
        3. Multiple Inheritance
        4. Hierarchical Inheritance
        5. Hybrid Inheritance - Hybrid is a combination od mutilple and hierarchical inheritance and due to it's shape of presentation it is also known as diamond inheritance.
</pre>

```python
# Single Inheritance:
class A:   # Parent class
    def add(self,x,y):
        print("Parent class A's add():",x+y)

class B(A):
    def sub(self,m,n):
        print("Intermediate Parent class B's sub():",m-n)

class C(B):
    def mod(self,p,q):
        print("Child class's C mod():",p%q)

ob=C()
ob.add(23,10)
ob.sub(30,10)
ob.mod(22,12)
```

```python
ob1 = B()
print(ob1)
ob1.add(12,38)
ob1.sub(34353,545)
```

```python
# Multi-level Inheritance: It consists of one parent class, one child class and multiple intermediate parent classes.
# A single child class.
class X:
    def display(self):
        print("Parent class X's display method.")
class Y:
    def show(self):
        print("Parent class Y's show method.")
class Z(X,Y):
    def present(self):
        print("Child class Z's present method.")

ob = Z()
ob.display()
ob.show()
ob.present()
```

```python
# Hierarchical Inheritance:
class X:
    def display(self):
        print("Parent class X's display method.")
class Y(X):
    def show(self):
        print("Parent class Y's show method.")
class Z(X):
    def present(self):
        print("Child class Z's present method.")
        
x = Y()
x.display()
x.show()
y = Z()
y.display()
y.present()
```

```python
# Hybrid Inheritance:
class A:   # Parent class
    def add(self,x,y):
        print("Parent class A's add():",x+y)

class B(A):
    def sub(self,m,n):
        print("Intermediate Parent class B's sub():",m-n)

class C(A):
    def mod(self,p,q):
        print("Intermediate Parent class's C mod():",p%q)

class D(B,C):
    def div(self,i,o):
        print("Child class's D div():",i/o)

x = D()
x.add(34,10)
x.sub(10,5)
x.mod(5,8)
x.div(50,10)
```

```python
# Create a class polygon that receives length of the sides and have a method to calculate the perimeter.
class Polygon:
    def __init__(self, *sides):
        self.sides = sides
        print('Total Polygon Sides:',self.sides)

    def Perimeter(self):
        print('Perimeter:',sum(self.sides))

p = Polygon(1,2,3)
p.Perimeter()
```

```python
# Create another class Triangle that inherits from polygon class and have only method to calculate the area of the triangle.
class Triangle(Polygon):
    def Area(self):
        s = self.Perimeter()/2
        area = (s*(s-self.sides[0])*(s-self.sides[1])*(s-self.sides[2]))**0.5
        print("Area of the triangle:",area)

tri = Triangle(1,2,3)
tri.Area()
```

## Encapsulation:

### Encapsulation means wrapping up of all the data into a single unit.
<pre>
    <b> 
        1. Class itself is an example of encapsulation.
        2. In python encapsulation means adding restrictions to the accessibility of the members.

        
        Encapsulation is divided into two types:
        
        -> Private Attribute: 
        
        If a variable is starting with double underscore it is treated as Private Attribute.

        1. A private member cannot be accessed by any object.
        2. Cannot be accessed by a class directly.
        3. A private member can be only accessed by a method

        
        -> Protected Attribute: 
        
        If a variable is staring with single underscore it is treated as Protected Attrtibute.

        1. Can be accessed by an object.
        2. Can be accessed by a class.
        3. Can be accessed by a method.

        
</pre>

```python
class Encapsulate:
    _var1 = 12345678
    __var2 = 98765432
    def show(self):
        print("Can access a protected member:", self._var1)
        print("Can access a private member:", self.__var2)
```

### (Poly - many, morphi - forms) In python polymorphism means having multiple methods in different classes or same class but acting differently everywhere.

<pre>
    <b>
        Polymorphism are of two types:

        1. Run-time Polymorphism:
        
 * Run-Time Polymorphism is also known as method overriding. It can only take place in the concept of inheritance.
 * In run-time polymorhism the methods present in the parent class and the child class will have same name and the arguments can be same or different.
        
        2. Compile-time Polymorphism:
 
 * Compile-Time Polymorphism is also known as Method Overloading.
 * It can be achieved within a single class having multiple methods.
 * In method overloading the multiple methods present in a single class will have same name and the arguments will be different.
</pre>

## Method Overriding / Runtime Polymorphism

```python
# Write the function to calculate the sum of digits of a positive integer.
def sum_of_digits(n):
    total = 0
    while n > 0:
        digit = n % 10  # Get the last digit
        total += digit   # Add it to the total
        n //= 10         # Remove the last digit
    return total

# Example usage
number = int(input("Enter a positive integer: "))
print(f"The sum of digits of {number} is {sum_of_digits(number)}")
```

```python
# Q - Create a class named Rectangle with the attributes length, width and methods to calculate area and perimeter
class Rectangle:
    def __init__(self, length=3, width=4):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

rect = Rectangle()
print("Area:", rect.area())
print("Perimeter:", rect.perimeter())

    
```

```python
# Pg - 16 Q - 4:
class Car:
    Company = 'Tata'
    def __init__(self,brand,model,fuel_percent,km_driven):
        self.brand = brand
        self.model = model
        self.fuel_percent = fuel_percent
        self.km_driven = km_driven

    def Display(self):
        print("Company:",self.Company)
        print("Brand:",self.brand)
        print("Model:",self.model)
        print("Fuel Percent:",self.fuel_percent)
        print("Kilometers Driven:",self.km_driven)

    def Fill(self,amount):
        self.fuel_percent += amount
        print("Updated Fuel Percent:",self.fuel_percent)

    def Drive(self,km):
        self.km_driven+= km
        self.fuel_percent -= 10
        self.Display()

ob = Car('Land Rover','Range Rover',67,100)
ob.Fill(70)
ob.Drive(75)
```
